In a GeoPackage (`.gpkg`) file, vector data can be stored in a variety of ways as standardised by the OGC. In this example the **country boundary** (`landgebied`), **province boundary** (`provinciegebied`), and **municipality boundary** (`gemeentegebied`) layers are all of the `MultiPolygon` geometry type, i.e. containing more than one polygon.

## Opening Vector Files as GeoDataFrames

You can open a layer of a GeoPackage as an individual GeoDataFrame using the GeoPandas library's `.read_file()` function. You have to provide the file path and if required the layer name.

````python
country_gdf = gpd.read_file(raw_data_path, layer='landgebied')
provinces_gdf = gpd.read_file(raw_data_path, layer='provinciegebied')
cities_gdf = gpd.read_file(raw_data_path, layer='gemeentegebied')
````

Here, `landgebied`, `provinciegebied`, `gemeentegebied` are the three individual layers of our GeoPackage opened as three different GeoDataFrames with GeoPandas' `gpd.read_file()` function. 

Then, `country_gdf`, `province_gdf`, `city_gdf` are three assigned GeoDataFrames to the three variable names respectively.

````python
provinces_gdf = provinces_gdf.astype({
    'code': 'int64',
    'ligt_in_land_code': 'int64'
})

provinces_gdf.dtypes
````

````python
country_gdf = country_gdf[['naam', 'geometry']]
provinces_gdf = provinces_gdf[['naam', 'geometry']]
cities_gdf = cities_gdf[['naam', 'geometry']]

print(f'Country: {country_gdf.columns}')
print(f'Provinces: {provinces_gdf.columns}')
print(f'Municipalities: {cities_gdf.columns}')
````

````python
provinces_gdf['naam'] = provinces_gdf['naam'].fillna('Unknown')

print('Filled missing values')
````

````python
provinces_gdf = provinces_gdf.dropna(subset=['naam', 'geometry'])

print('Dropped missing values')
````

````python
provinces_gdf.isna().sum()
````

````python
country_gdf = country_gdf[
    country_gdf.geometry.notna() & 
    ~country_gdf.geometry.is_empty
]

provinces_gdf = provinces_gdf[
    provinces_gdf.geometry.notna() & 
    ~provinces_gdf.is_empty
]

cities_gdf = cities_gdf[
    cities_gdf.geometry.notna() & 
    ~cities_gdf.is_empty
]

print('Dropped rows with null and empty geometry values')
````

````python
country_gdf = country_gdf.rename(columns={
    'naam': 'name'
})

provinces_gdf = provinces_gdf.rename(columns={
    'naam': 'name'
})

cities_gdf = cities_gdf.rename(columns={
    'naam': 'name'
})

print(f'Country:{country_gdf.columns}')
print(f'Provinces: {provinces_gdf.columns}')
print(f'Municipalities: {cities_gdf.columns}')
````

````python
country_gdf['geometry'] = country_gdf['geometry'].make_valid()
provinces_gdf['geometry'] = provinces_gdf['geometry'].make_valid()
cities_gdf['geometry'] = cities_gdf['geometry'].make_valid()

print('Geometry validation completed')
````

````python
country_gdf = country_gdf.to_crs('EPSG:28992')
provinces_gdf = provinces_gdf.to_crs('EPSG:28992')
cities_gdf = cities_gdf.to_crs('EPSG:28992')

print('Reprojection completed')
````

````python
print(f'Country: {country_gdf.crs}')
print(f'Provinces: {provinces_gdf.crs}')
print(f'Municipalities: {cities_gdf.crs}')
````

````python
gdf_export_path = Path('../data/outputs/netherlands.gpkg')

country_gdf.to_file(gdf_export_path, layer='country', driver='GPKG', mode='w')
provinces_gdf.to_file(gdf_export_path, layer='provinces', driver='GPKG', mode='a')
cities_gdf.to_file(gdf_export_path, layer='cities', driver='GPKG', mode='a')

print('Exported files')
````
