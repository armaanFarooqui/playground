# Getting Started
## Virtual Environments

### Creating a Virtual Environment

* Navigate to your project's root directory using the following shell command. Replace the example path with the actual location of your project.

```shell
  cd ~/Projects/geoproject
```


* Now, create a Python virtual environment.

  ```shell
  python3 -m venv .venv
  ```

Here,

`python3` invokes the Python interpreter, which executes the instructions that follow.

`-m` tells Python to execute a specified module as a program.

`venv` is a module included in Python's standard library that creates virtual environments.

`.venv` specifies the name of the directory in which the virtual environment will be created. The dot (`.`) at the beginning makes the directory hidden by default on Linux and other Unix-like operating systems.

It is highly recommended to create a separate virtual environment for each Python project. This allows project-specific Python packages to be installed without modifying the system-wide Python installation or interfering with packages required by other projects.

Remember to delete your existing virtual environment and recreate it anew when you rename your move your project root folder. You will also have to reinstall your Python depencies using the `requirements.txt` file which is discussed below.

### Entering a Virtual Environment

* Activate your virtual environment using the following shell command.

  ```shell
  source .venv/bin/activate
  ```

You will need to execute this command whenever you open a new terminal session and want to work inside this virtual environment.

Here,

`source` is a shell command that executes the contents of a script.

`.venv/bin/activate` specifies the path to the activation script. Executing this script activates your virtual environment.

### Exiting a Virtual Environment

* To deactivate the virtual environment, execute:

  ```shell
  deactivate
  ```

This restores the shell's previous environment settings.

## Python Dependencies

### Documenting Python Dependencies

In this guide, we will use several Python packages. A `requirements.txt` file allows us to document the Python packages required by our project and install them again in the future. The filename `requirements.txt` is a convention rather than a requirement.

Make sure that your terminal is in your project's root directory. If necessary, use the `cd` shell command to navigate to the correct location. For example:

```shell
cd ~/Projects/geoproject
```

* Create an empty text file using the following shell command:

  ```shell
  > requirements.txt
  ```

Here, `>` is a shell redirection operator that creates an empty file if it does not already exist. If the file already exists, its contents are deleted.

`requirements.txt` specifies the path and filename of the file.

* You should now see a file named `requirements.txt` in your file manager. Open this file using a text editor.

* Make sure the file is empty. Paste the following package names into it and save the file.

  ```text
  notebook
  numpy
  pandas
  matplotlib
  geopandas
  xarray
  rioxarray
  ```

Here, the following Python packages have been specified:

`notebook` for working with Jupyter notebooks (`.ipynb` files).

`numpy` for working with numerical data.

`pandas` for analysing tabular data.

`matplotlib` for visualising data.

`geopandas` for working with geospatial vector data.

`xarray` for working with multidimensional data.

`rioxarray` for working with geospatial raster data.

You should document the Python packages required by your project in this file. This makes it easier to recreate the project's Python environment.

### Installing Documented Python Dependencies

We can use a `requirements.txt` file to install the Python packages required by our project.

Make sure that your terminal is in your project's root directory. If not, navigate to it using the `cd` command. For example:

```shell
cd ~/Projects/geoproject
```

Make sure that your Python virtual environment is activated. If not, activate it using the `source` command and the path to its activation script. For example:

```shell
source .venv/bin/activate
```

* Now, install the Python packages specified in your `requirements.txt` file using the following command:

```shell
python3 -m pip install -r requirements.txt
```

Here,

`python3` invokes the Python interpreter associated with the activated virtual environment.

`-m` tells Python to execute a specified module as a program.

`pip` is Python's package installer used to install packages from the Python Package Index (PyPI).

`install` is a pip command that installs the specified packages.

`-r` tells pip to read package requirements from a specified file.

`requirements.txt` specifies the path to your `requirements.txt` file.

Pip will install the packages listed in this file, along with any additional Python packages they require.

## Jupyter Notebook

### Launching Jupyter Notebook Server

Make sure that your terminal is in your project's root directory. If not, navigate to it using the `cd` command. For example:

```shell
cd ~/Projects/geoproject
```

Make sure that your Python virtual environment is activated. If not, activate it using the `source` command and the path to its activation script. For example:

```shell
source .venv/bin/activate
```

* Now, launch Jupyter Notebook using the following command:

```shell
jupyter notebook
```

This command starts a local Jupyter server, typically on port `8888`. The Jupyter Notebook interface connects to this server, allowing you to work with notebooks (`.ipynb` files).

The interface should automatically open in your default web browser. Keep the terminal running, as closing it terminates the Jupyter server.

### Closing the Jupyter Notebook Server

* To stop the Jupyter Notebook server, press `Ctrl` + `C` in the terminal where the server is running.

* If the terminal asks you to confirm the shutdown, press `Y` followed by `Enter`.

The Jupyter server will then shut down.
